```python
import pandas as pd

print("Pandas version:", pd.__version__)

df = pd.read_excel('AHA_1300_Database.xlsx')

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
# Display all column names
print("Dataset columns:")
print(df.columns.tolist())

# Display information about the dataset
print("\nDataset information:")
df.info()

# Display missing values in each column
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# Compare existing indicator against the Births column
expected = (df['Births'] > 0).astype(int)

# Find inconsistencies among existing, non-missing values
mismatches = df[
    df['Births or Not'].notna() &
    (df['Births or Not'] != expected)
]

print("Existing records that disagree with the Births indicator:")
print(mismatches[
    ['Number', 'Births', 'Births or Not']
])

print("\nNumber of inconsistencies:", len(mismatches))
```

```python
# Reconstruct Births or Not using the Births column
df['Births or Not'] = (df['Births'] > 0).astype(int)

# Verify the reconstructed column
print("Reconstructed Births or Not:")
print(df['Births or Not'].value_counts())

print("\nMissing values after cleaning:")
print(df.isnull().sum())

# Check hospital 353
print("\nCorrected hospital 353:")
print(df.loc[df['Number'] == 353,
             ['Number', 'Births', 'Births or Not']])
```

```python
# Export the cleaned dataset to Excel
df.to_excel('AHA_1300_Database_Cleaned.xlsx', index=False)

print("Cleaned dataset exported successfully!")
```
