```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the pre-cleaned dataset
df = pd.read_csv('../data/processed/aha_cleaned.csv')
print(f"Loaded {df.shape[0]} rows and {df.shape[1]} columns.")
```

```python

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Admissions",
    y="Total Expense"
)

plt.title("Relationship Between Hospital Admissions and Total Expenses")
plt.xlabel("Number of Admissions")
plt.ylabel("Total Expenses")
plt.grid(True)

plt.savefig(
    "../outputs/figures/admissions_vs_total_expense.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

correlation = df["Admissions"].corr(df["Total Expense"])
print("Pearson correlation:", correlation)
```


### Scatter Plot: Admissions vs Total Expense

The scatter plot shows a strong positive relationship between hospital admissions and total expenses.

The Pearson correlation coefficient was 0.8744, indicating a strong positive linear relationship. Hospitals with higher admissions generally tend to have higher total expenses.

However, correlation does not necessarily imply causation.
