```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the pre-cleaned dataset
df = pd.read_csv('../data/processed/aha_cleaned.csv')
print(f"Loaded {df.shape[0]} rows and {df.shape[1]} columns.")
```

```python
variables = ['Beds','Admissions','Census']
results = df[variables].agg(
    ['mean','median','min','max','std']
).T
results['Range'] = results['max'] - results['min']

print(results)
```

```python
results.round(2).to_csv('../outputs/tables/02_continuous_metrics.csv', index=True)
```
